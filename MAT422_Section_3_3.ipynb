# MAT 422 - Section 3.3: Optimization and Gradient Descent
**Topics**: Necessary and sufficient conditions of local minimizers, Convexity and global minimizers, Gradient descent

```python
import numpy as np
import matplotlib.pyplot as plt
```

# 3.3.1 Necessary and Sufficient Conditions of Local Minimizers

```python
def f(x):
  return x**3 - 3*x
def f_prime(x):
  return 3*x**2 - 3
def f_prime_prime(x):
  return 6*x

critical_points = np.array([-1, 1])
for c in critical_points:
    print(f"x={c}: f'(x)={f_prime(c)}, f''(x)={f_prime_prime(c)}, f(x)={f(c)}")

x = np.linspace(-2, 2, 400)
plt.plot(x, f(x))
plt.scatter(critical_points, f(critical_points), color="tab:red")
plt.title("Local Minimizer and Maximizer")
plt.show()
```

My Explanation: When you take the derivative of a function, you can find the critical points where the derivative is equal to zero. With those critical points, you can find the classification of the points, whether or not they are local minimizers or maximizers, which are valleys or peaks of the original function in a local range.

# 3.3.2 Convexity and Global Minimizers

```python
def g(x):
  return (x + 1)**2 + 3

x = np.linspace(-3, 1, 400)
plt.plot(x, g(x))
plt.axvline(-1, color="tab:red", linestyle="--", label="Global Minimizer")
plt.title("Global Minimizer of $g(x)$")
plt.show()
```

My Explanation: For any given convex function, we know that a local minimizer is actually the global minimizer for the function. Because the chosen function is convex, the graph shows that it has a minimizer at $x = -1$ and we can say that it is a global minimizer.

# 3.3.3 Gradient Descent

```python
def gradient_g(x):
  return 2*x + 2

learning_rate = 0.2
current = 2.0
path = [current]
for _ in range(20):
    current = current - learning_rate * gradient_g(current)
    path.append(current)

path = np.array(path)
print("Final x =", path[-1])
print("Final objective =", g(path[-1]))

plt.plot(path, marker="o")
plt.axhline(-1, color="tab:red", linestyle="--", label="Minimizer")
plt.xlabel("iteration")
plt.ylabel("x value")
plt.title("Iteration of Gradient Descent")
plt.legend()
plt.show()
```

My Explanation: Gradient descent is an interative optimization algorithm that can be used to find a local minimizer of a function. Starting with a random point, the gradient is subtracted iteratively until the minimizer is approached. This can be shown in the graph, as the minimizer is at $-1$ and the $x$ starts at $2$ before decreasing and approaching $-1$, the true minimizer.
