```python
import numpy as np
import matplotlib.pyplot as plt
import math
from IPython.display import Markdown

np.set_printoptions(precision=4, suppress=True)
```

# 2.2.1 - Probability Axioms

```python
S = {1, 2, 3, 4, 5, 6}
A = {2, 4, 6}
B = {4, 5, 6}

S_probability = len(S) / len(S)
display(Markdown(r"Probability of Any Outcome ($P(S)$):"))
display(S_probability)

A_probability = len(A) / len(S)
display(Markdown(r"Probability of Event A ($P(A)$):"))
display(A_probability)

B_probability = len(B) / len(S)
display(Markdown(r"Probability of Event B ($P(B)$):"))
display(B_probability)

intersection = A & B
intersection_probability = len(intersection) / len(S)
display(Markdown(r"Probability of Event A and B ($P(A \cap B)$):"))
display(intersection_probability)

union = A | B
union_probability = len(union) / len(S)
display(Markdown(r"Probability of Event A or B ($P(A \cup B)$):"))
display(union_probability)
display(Markdown(r"Probability of Event A plus B minus A and B ($P(A) + P(B) - P(A \cap B)$):"))
display(A_probability + B_probability - intersection_probability)

complement = S - A
complement_probability = len(complement) / len(S)
display(Markdown(r"Probability of Event A' ($P(A^c)$):"))
display(complement_probability)
display(Markdown(r"Probability of Event A plus A' ($P(A) + P(A^c)$):"))
display(A_probability + complement_probability)

display(Markdown(r"Mutually Exclusive Events ($P(A) + P(A^c)$):"))
display(A_probability + complement_probability)
```

My Explanation: Probability has a handful of axioms that are be displayed here. We have $S$, the different outcomes of rolling a die, $A$, the even outcomes, and $B$, the outcomes higher than three. The probability of getting something in the set is 1 because no matter the outcome, it will be one of the six sides of the die. Because both $A$ and $B$ have three of the six outcomes of the die, they have a probability of 0.5. The intersection shows the probability of the outcomes in both A and B, whereas the union shows the probability of the outcomes in either A or B. The complement is the opposite of a set and when added with the original set equals 1. We can also see that mutually exclusive probabilities are additive. Finally, while implied, all of the probabilites are positive.

# 2.2.2 - Conditional Probability

```python
conditional_probability_formula = intersection_probability / B_probability
display(Markdown(r"Conditional Probability Using Formula ($P(A|B)$):"))
display(conditional_probability_formula)

conditional_probability_sample_space = len(intersection) / len(B)
display(Markdown(r"Conditional Probability Using Sample Space ($P(A|B)$):"))
display(conditional_probability_sample_space)

multiplication_rule = conditional_probability_formula * B_probability
display(Markdown(r"Multiplication Rule ($P(A|B)*P(B)$):"))
display(multiplication_rule)
display(Markdown(r"Does Multiplication Rule Equal Intersection? ($P(A \cap B) =?= P(A|B)*P(B)$)"))
display(intersection_probability == multiplication_rule)

independence = conditional_probability_formula == A_probability
display(Markdown(r"Independence ($P(A|B) =?= P(A)$)"))
display(independence)
```

My Explanation: Conditional Probability is the probability that something will occur once something else has happened. For example, $P(A|B)$ is the probability that $A$ occurs if $B$ has already occurred. This changes the possible outcomes of $A$ from {1, 2, 3, 4, 5, 6} to {4, 5, 6}, which results in the $\frac{2}{3}$ probability instead of $\frac{1}{2}$. The multiplication rule is another way to find the probability of the intersection using the conditional probability. That is checked and proven true. Finally, independence between two variables shuold result in the conditional probability just being equal to the probability of the first. In this case, $A$ and $B$ are not independent, thus their independence is shownn as false.

# 2.2.3 - Discrete Random Variables

```python
#Probability Mass Function
X = [0, 1, 2, 3, 4, 5]
n = 5
p = 0.5
def binomial_pmf(n, p, x):
    return math.comb(n, x) * p**x * (1-p)**(n-x)
for x_val in X:
  probability = binomial_pmf(n, p, x_val)
  print(f"P({x_val}) = {probability}")
total = sum([binomial_pmf(n, p, x_val) for x_val in X])
print(f"Total = {total}")

#Cumulative Distribution Function
at_most_three_heads_probability = sum([binomial_pmf(n, p, x_val) for x_val in range(4)])
display(Markdown(r"Proability of at most three heads ($P(X <= 3)$)"))
display(at_most_three_heads_probability)

#Expected Value
expected_value = sum([x_val * binomial_pmf(n, p, x_val) for x_val in X])
display(Markdown(r"Expected Value ($E(X)$)"))
display(expected_value)

#Variance
variance = sum([(x_val - expected_value)**2 * binomial_pmf(n, p, x_val) for x_val in X])
display(Markdown(r"Variance ($Var(X)$)"))
display(variance)

#Standard Deviation
standard_deviation = math.sqrt(variance)
display(Markdown(r"Standard Deviation ($SD(X)$)"))
display(standard_deviation)
```

My Explanation: Discrete Random Variables show the possibilites of a finite number of outcomes of a test. For my example, you can flip an coin five times and count the number of times it lands on heads. This will give you the possibilities {0, 1, 2, 3, 4, 5}, which is $X$. We can use the PMF to calculate the chance that the exact number of heads occurs. The CDF can be used to figure out the probability of the situation occuring up to a certain number of times. In my case, I chose to find the probability that no more than three heads occur. The expected value is the expected amount of times that the process will succeed giving the amount of tests and probability. The variance and standard deviation both show how spread out the options are from the mean.

# 2.2.4 Continous Random Variables

```python
from scipy.integrate import quad

#Normal Probability Density Function
mu = 70
sigma = 10
def normal_pdf(x, mu, sigma):
    return (1 / (sigma * math.sqrt(2 * math.pi))) * math.exp(-((x - mu)**2) / (2 * sigma**2))
normal_pdf_probability, error = quad(lambda x: normal_pdf(x, mu, sigma), 60, 80)
display(Markdown(r"Probability Density Function ($P(60 <= X <= 80)$)"))
display(normal_pdf_probability)

normal_pdf_total = quad(lambda x: normal_pdf(x, mu, sigma), -math.inf, math.inf)
display(Markdown(r"Total Probability ($P(X)$)"))
display(normal_pdf_total)

normal_pdf_expected_value = quad(lambda x: x * normal_pdf(x, mu, sigma), -math.inf, math.inf)
display(Markdown(r"Expected Value ($E(X)$)"))
display(normal_pdf_expected_value)

normal_pdf_variance = quad(lambda x: (x - mu)**2 * normal_pdf(x, mu, sigma), -math.inf, math.inf)
display(Markdown(r"Variance ($Var(X)$)"))
display(normal_pdf_variance)

normal_pdf_standard_deviation = math.sqrt(normal_pdf_variance[0])
display(Markdown(r"Standard Deviation ($SD(X)$)"))
display(normal_pdf_standard_deviation)

normal_pdf_standardization = (80 - mu) / sigma
display(Markdown(r"Standardization ($Z$):"))
display(normal_pdf_standardization)
```

My Explanation: The Normal Probability Density Function is an example of the continous random variables. For my example, I chose $\mu = 70$ and $\sigma = 10$ for a Normal PDF of $X$ ~ $N(70,10^2)$. Instead of finding the probability of one specific case, this finds the area under the curve of a probability density curve. Integrating it from 60 to 80 gives a probability of 0.682 or 68.2%, meaning that 68.2% of the distribution falls within that range. Because 60 and 80 are both one standard deviation away from the mean, this shows the probability of being within one standard deviation of the mean. Integrating the PDF across all values gives 1, showing that the probability of the entire distribution is 100%. The expected value is found by integrating $xf(x)$, the variance is found from integrating $(x-\mu)^2f(x)$, and taking the square root of the variance gives the standard deviation. Finally, standarization uses $Z = (X - \mu)/\sigma$ to describe how many standard deviations a value is away from the mean. For $X=80$, the z score is one, meaning it is one standard deviation above the mean.
