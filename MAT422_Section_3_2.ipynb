```python
import numpy as np
import matplotlib.pyplot as plt
import math
from IPython.display import Markdown
```

#3.2.1 Limits and Continuity

```python
h = np.array([1, 0.5, 0.1, 0.01, 0.001])
values = (np.exp(h) - 1) / h
print("h values:", h)
display(Markdown(r"$\frac{e^h-1}{h}$ values:"))
print(values)

x = np.linspace(-1, 3, 400)
y = np.where(x < 1, x + 2, x**2 + 3)
plt.plot(x,y)
plt.axvline(0, color="black", linewidth=0.8)
plt.title("Example of a jump discontinuity at x = 1")
plt.show()
```

My Explanation: The limit of a function can show what value a function approaches as the input gets closer to a specific number. In my example, we are seeing what value the function $\frac{e^h-1}{h}$ approaches as $h\rightarrow0$. We can see that as h decreases to 0, $\frac{e^h-1}{h}$ approaches 1. This can be important when the specific value of the function does not exist, like in this case.
Continuity can be shown through limits as well. With the piecewise function $y = \begin{cases} x+2 & \text{if } x < 1 \\ x^2 + 3 & \text{if } x \ge 1 \end{cases}$ we can see that $\lim_{x \rightarrow 1^-} x + 2 = 3$, but $\lim_{x \rightarrow 1^+} x^2 + 3 = 4$, showing that there will be a jump discontinuity when $x \rightarrow 1$.

#3.2.2 Derivatives

```python
def f(x):
    return x**4 - 3*x**2 + 2*x

point = 3
steps = np.array([1e-1, 1e-2, 1e-3, 1e-4])
numeric_derivatives = (f(point + steps) - f(point - steps)) / (2 * steps)
exact_derivative = 4*point**3 - 6*point + 2

print("Numeric Derivative Estimates:", numeric_derivatives)
print("Exact Derivative:", exact_derivative)
print("Errors:", np.abs(numeric_derivatives - exact_derivative))
```

My Explanation: The derivative of a function can be found by the limit definition of the derivative, $\lim_{h \rightarrow 0} \frac{f(x-h) - f(x)}{h}$ or by exact formula. The formula for the derivative of $x^4 - 3x^2 + 2x$ is $4x^3 - 6x + 2$. We can calculate both by approximating the limit as $h$ approaches 0 as well as taking the exact derivative. This explains the error between the numeric estimations and the exact derivative.

#3.2.3 Taylor's Theorem

```python
x = np.linspace(-2, 2, 300)
taylor_sin_x = x - x**3 / 6 + x**5 / 120

plt.plot(x, np.sin(x), label="$sin(x)$")
plt.plot(x, taylor_sin_x, "--", label="third-degree Taylor polynomial at 0")
plt.xlabel("x")
plt.ylabel("value")
plt.title("Taylor approximation")
plt.legend()
plt.show()

print("Max error on [-2, 2] =", np.max(np.abs(np.sin(x) - taylor_sin_x)))
```

My Explanation: Taylor's Theorem states that any function that can be differentiated $n$ times has a polynomial approximation up to the $nth$ degree with a small error. This estimation can be seen in the graph comparing the function $sin(x)$ to the Taylor Series Approximation $x - \frac{x^3}{3!} + \frac{x^5}{5!}$. The error between this estimation and the function itself is then given.
